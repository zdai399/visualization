```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
df = pd.read_csv("/Users/ziyidai/Downloads/Licensed Dogs and Cats.csv") #import dataset
print(df)
```

```python
pet_counts_per_year= (df.groupby("Year")
                      .size()
                      .reset_index(name= "num_pets")
                      .sort_values("Year")) #I made a dataset counting the number of pets per year 

plt.figure(figsize=(10,3))
bargraph= sns.barplot(data=pet_counts_per_year, #chose the dataset
x="Year", #defined x variable
y= "num_pets" #defined y variable
)
bargraph.set(title='Number of licensed dogs and cats in the city of Toronto from 2023-2026(Jan.31st)',
xlabel='Year',
ylabel='Number of pets')
```
